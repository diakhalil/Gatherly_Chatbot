# Document Extraction Experiments


## 1. Setup

```python
import importlib
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
SRC_DIR = PROJECT_ROOT / "src"

if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

PDF_FILES = [
    
    PROJECT_ROOT / "docs" / "Organiser_un_evenement_deAaZ.pdf",
    PROJECT_ROOT / "docs" / "Checklists.pdf",
    PROJECT_ROOT / "docs" / "Guideline_Sustainable_Event.pdf",
    PROJECT_ROOT / "docs" / "تقاليد الزفاف في الثقافات العربية.pdf",
]

print("Project root:", PROJECT_ROOT)

for pdf_file in PDF_FILES:
    print("\nPDF:", pdf_file)
    print("Exists:", pdf_file.exists())


    
THEMES_PDFS_DIR = (
    PROJECT_ROOT
    / "docs"
    / "generated_pdfs"
)

themes_pdfs = sorted(
    THEMES_PDFS_DIR.glob("*.pdf")
)

PDF_FILES.extend(themes_pdfs)
```

## 2. Load and Inspect the Test Document

### 2.1 Validate the Document and Detect Its Type

```python
loader_module = importlib.import_module(
    "doc_rag.1_loader"
)

detect_document_type = (
    loader_module.detect_document_type
)

load_document = (
    loader_module.load_document
)

document_infos = [
    load_document(pdf_file)
    for pdf_file in PDF_FILES
]

for document_info in document_infos:
    print("\nDocument:", document_info["file_name"])
    print("Type:", document_info["file_type"])
    print("Size:", document_info["file_size_bytes"], "bytes")
    print("Path:", document_info["source_path"])
```

### 2.2 Extract Native PDF Content

```python
extraction_module = importlib.import_module(
    "doc_rag.2_extraction"
)

extract_pdf = (
    extraction_module.extract_pdf
)

extracted_documents = [
    extract_pdf(pdf_file)
    for pdf_file in PDF_FILES
]

for extracted_document in extracted_documents:
    print("\nFile:", extracted_document["file_name"])
    print("Type:", extracted_document["file_type"])
    print("Pages:", extracted_document["page_count"])
    print("Metadata:", extracted_document["metadata"])
    
```

## 3. Build the Page Inventory

```python
import pandas as pd

page_inventory = pd.DataFrame([
    {
        "file_name": document["file_name"],
        "page_number": page["page_number"],
        "text_length": page["text_length"],
        "text_blocks": len(page["text_blocks"]),
        "width": page["width"],
        "height": page["height"],
        "native_text_usable": page["native_text_usable"],
        "ocr_text_length": page["ocr_text_length"],
    }
    for document in extracted_documents
    for page in document["pages"]
])

page_inventory
```

## 4. Compare Extraction Coverage

```python
page_inventory.groupby("file_name")[
    [
        "text_length",
        "text_blocks",
    ]
].describe()
```

## 5. Select Representative Pages

```python
for file_name, document_pages in page_inventory.groupby(
    "file_name",
    sort=False,
):
    print(f"\nDOCUMENT: {file_name}")

    print("\nPages with most native text:")
    display(
        document_pages.sort_values(
            "text_length",
            ascending=False,
        ).head(5)
    )

    print("Pages with least native text:")
    display(
        document_pages.sort_values(
            "text_length",
            ascending=True,
        ).head(5)
    )
```

## 6. Inspect Native Text and Reading Order

```python
for extracted_document in extracted_documents:
    document_inventory = page_inventory[
        page_inventory["file_name"]
        == extracted_document["file_name"]
    ]

    text_heavy_page_number = int(
        document_inventory.sort_values(
            "text_length",
            ascending=False,
        ).iloc[0]["page_number"]
    )

    text_heavy_page = extracted_document["pages"][
        text_heavy_page_number - 1
    ]

    print("\n" + "=" * 80)
    print("DOCUMENT:", extracted_document["file_name"])
    print("PAGE:", text_heavy_page_number)
    print(
        "TEXT BLOCKS:",
        len(text_heavy_page["text_blocks"]),
    )
    print("\nEXTRACTED TEXT\n")
    print("EXTRACTION METHOD:", text_heavy_page["extraction_method"])
    print()
    print(text_heavy_page["final_text"])
```

```python
target_document = next(
    document
    for document in extracted_documents
    if document["file_name"] == "Wedding_Themes.pdf"
)

target_page = next(
    page
    for page in target_document["pages"]
    if page["page_number"] == 4
)

print("TEXT LINES\n")

for index, line in enumerate(
    target_page["text_lines"],
    start=1,
):
    print(
        f"{index:02d} | "
        f"bbox={line['bbox']} | "
        f"{line['text']}"
    )
```

### checking 'checklists' ocr

```python
import json

pages_to_show = [ 4, 5, 6, 8]  

checklist_dir = next(Path(PROJECT_ROOT / "data/rag/processed_documents").glob("Checklists__*"))
pages = {
    p["page_number"]: p
    for p in json.loads((checklist_dir / "extracted_text.json").read_text(encoding="utf-8"))["document"]["pages"]
}

for n in pages_to_show:
    page = pages[n]
    text = (page.get("final_text") or "").strip()
    if not text:
        print(f"PAGE {n}: (empty)\n")
        continue
    print(f"PAGE {n} - {page['extraction_method']} - len={len(text)}")
    print(text)
    print("\n" + "=" * 72 + "\n")
```

### testing diff arabic ocr

```python
import subprocess
import tempfile
from pathlib import Path

import pymupdf
```

```python
arabic_pdf = PROJECT_ROOT / "docs" / "تقاليد الزفاف في الثقافات العربية.pdf"

page_number = 5

with pymupdf.open(arabic_pdf) as pdf:
    page = pdf.load_page(page_number - 1)

    pix = page.get_pixmap(
        dpi=300,
        alpha=False,
    )

    temp_image = Path(
        tempfile.gettempdir()
    ) / "arabic_ocr_test.png"

    pix.save(temp_image)


for psm in [3, 4, 6, 11, 12]:

    
    print(f"PSM {psm}")
   

    result = subprocess.run(
        [
            "tesseract",
            str(temp_image),
            "stdout",
            "-l",
            "ara+eng",
            "--psm",
            str(psm),
        ],
        capture_output=True,
        text=True,
        encoding="utf-8",
        errors="replace",
    )

    print(result.stdout)
    print("\n\n\n")
```

## 7. Inspect Automatic Boilerplate Detection

Inspect the repeated header, footer, and navigation text detected
automatically across each document before chunking.

```python
for extracted_document in extracted_documents:

    print("\n" + "=" * 80)
    print(
        "DOCUMENT:",
        extracted_document["file_name"],
    )
    print("=" * 80)

    boilerplate = extracted_document[
        "boilerplate_lines"
    ]

    print(
        "DETECTED BOILERPLATE:",
        len(boilerplate),
    )

    if not boilerplate:
        print("None detected.")
        continue

    for line in boilerplate:
        print("•", repr(line))
```
